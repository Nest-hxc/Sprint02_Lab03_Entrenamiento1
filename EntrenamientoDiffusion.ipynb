```python
output_dir = "imagenes_generadas"
os.makedirs(output_dir, exist_ok=True)
```

```python
model_id = "stabilityai/stable-diffusion-2-1-base"
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
scheduler = EulerAncestralDiscreteScheduler.from_pretrained(
    model_id,
    subfolder="scheduler"
)
```

```python
pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    scheduler=scheduler
).to(device)
```

```python
prompts = [
    "A high-resolution product photo of modern running sneakers on a white background",
    "A luxury leather handbag with gold accents, studio lighting, white background"
]
```

```python
for i, prompt in enumerate(prompts):
    image = pipe(
        prompt,
        guidance_scale=7.5,
        num_inference_steps=30
    ).images[0]

    image.save(f"{output_dir}/producto_{i + 1}.png")

img2img_pipe = StableDiffusionImg2ImgPipeline.from_pretrained(
    model_id,
    scheduler=scheduler
).to(device)
```

```python
base_image = Image.open(f"{output_dir}/producto_1.png").convert("RGB")
base_image = base_image.resize((768, 768))

prompt_variacion = "modern running sneakers in red color with black details on a white background"
```

```python
new_image = img2img_pipe(
    prompt=prompt_variacion,
    image=base_image,
    strength=0.6,
    guidance_scale=7.5,
    num_inference_steps=30
).images[0]
```

```python
new_image.save(f"{output_dir}/producto_variacion.png")
```
